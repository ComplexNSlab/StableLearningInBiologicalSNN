```python
class FRNeuralNetwork:
    """ Standard framework of firing-rate neural networks """
    import numpy as np

    dt = 0.1
    eta = 0.01
    tau_x = 20
    tau = 50
    beta = 0.1
    
    def __init__(self, N):
        '''
            N : number of neurons in the network
        '''
        self.N = N
        
        # creation and initialization of features 
        
        # Connectivity Matrix
        self.W = 2*np.random.randn(N, N)/np.sqrt(N)

        # State Vector
        self.x = np.random.randn(N,1)


    def PHI():
        return np.tanh(self.x)
        
    def x_derivative(b):
        return -self.x + np.matmul(self.W, PHI()) + b
    
    def dynamic():
        
        self.x = self.x + x_derivative(b) * dt
```
