```python
from SmwusNetwork import SmwusNetwork
import matplotlib.pyplot as plt
import numpy as np

mynetwork = SmwusNetwork(state = 'Rate control')

dt = 0.1
T = 50000
save_w = []
save_x = []
sampling_rate = 100
time = np.arange(int(T/sampling_rate/dt))*dt*sampling_rate
for i in range(len(time)):
    mynetwork.update_network()
    save_w.append(mynetwork.W)
    save_x.append(mynetwork.x)

save_w = np.array(save_w)
save_x = np.array(save_x)

```

```python
for i in range(3):
    for j in range(3):
        plt.plot(time, save_w[:, i, j], label=str(i+1) + ',' + str(j+1))
plt.title(mynetwork.homeostasis_state)
plt.xlabel("time")
plt.ylabel("W_ij")
plt.legend()
plt.show()

for i in range(5):
    plt.plot(time[:], save_x[:, i], label=str(i+1))
plt.title(mynetwork.homeostasis_state)
plt.xlabel("time")
plt.ylabel("x_i")
plt.legend()
plt.show()
```
